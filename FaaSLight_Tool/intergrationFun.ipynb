```python
# 📌📌📌耗时久，不用重复执行，执行过一次就可以了，生成的output.json文件保存好就好
# 这里是对整个包进行静态分析，生成初始的调用关系图
# import os
print(input_package)
print(input_entry_point)
# os.system("python pycg/__main__.py --package {} {} -o {}".format(input_package, " ".join(input_entry_point), output_file))

import os
import subprocess
import sys

# 禁用调试器的模块验证（解决 imp 模块错误）
os.environ['PYDEVD_DISABLE_FILE_VALIDATION'] = '1'

print("input_package:", input_package)
print("input_entry_point:", input_entry_point)

# 构建命令
cmd = ["python3", "pycg/__main__.py", "--package", input_package] + input_entry_point + ["-o", output_file]

print("执行命令:", " ".join(cmd))

# 设置输出文件名
log_file = os.path.join(input_package, "pycg_complete_output.txt")

# 运行命令并同时写入文件和输出到控制台
with open(log_file, "w", encoding='utf-8') as log_f:
    # 写入命令信息
    log_f.write("="*80 + "\n")
    log_f.write("PyCG 完整执行日志\n")
    log_f.write("="*80 + "\n\n")
    log_f.write(f"命令: {' '.join(cmd)}\n\n")
    log_f.write("="*80 + "\n")
    log_f.write("STDOUT 输出:\n")
    log_f.write("="*80 + "\n")
    
    print("="*80)
    print("完整的 STDOUT 输出:")
    print("="*80)
    
    # 使用 subprocess.Popen 实时获取输出
    process = subprocess.Popen(
        cmd,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True,
        encoding='utf-8',
        bufsize=1,
        universal_newlines=True
    )
    
    # 实时读取 stdout
    stdout_lines = []
    for line in iter(process.stdout.readline, ''):
        line = line.rstrip('\n')
        if line:
            # 输出到控制台
            print(line)
            # 写入文件
            log_f.write(line + "\n")
            stdout_lines.append(line)
    
    # 等待进程结束
    process.wait()
    
    # 获取 stderr
    stderr_content = process.stderr.read()
    
    # 写入 stderr 到文件
    log_f.write("\n" + "="*80 + "\n")
    log_f.write("STDERR 输出:\n")
    log_f.write("="*80 + "\n")
    if stderr_content:
        log_f.write(stderr_content)
    else:
        log_f.write("(空)")
    
    # 写入返回码
    log_f.write("\n\n" + "="*80 + "\n")
    log_f.write(f"返回码: {process.returncode}\n")
    log_f.write("="*80 + "\n")

# 在控制台显示 stderr
print("\n" + "="*80)
print("完整的 STDERR 输出:")
print("="*80)
if stderr_content:
    print(stderr_content)
else:
    print("(空)")

print("\n返回码:", process.returncode)
print(f"\n✅ 完整输出已保存到: {log_file}")

# 检查输出文件
if os.path.exists(output_file):
    print(f"\n✅ 输出文件已创建: {output_file}")
    try:
        with open(output_file, 'r', encoding='utf-8') as f:
            content = f.read()
            print(f"文件大小: {len(content)} 字符")
            if len(content) > 0:
                print("文件内容预览:", content[:200])
    except Exception as e:
        print(f"读取输出文件时出错: {e}")
else:
    print(f"\n❌ 输出文件未创建")
```

```python
#(1) Function rewriting
# 代码重写与存根化。其核心目的是：把那些“没被用到的函数”从源代码中剔除，并替换成一个极其精简的“代理钩子”（Stub），从而实现 90% 以上的体积缩减。
# 
# 例如 numpy 库中的 mean 函数，如果它在运行过程中没有被调用到，那么它的源代码会被剔除，替换成如下的存根代码：
# import custom_funtemplate
# # 这里捕获了调用该函数所需的所有上下文
# return custom_funtemplate.rewrite_template(
#     'numpy.core.numeric.mean',           # 唯一标识符
#     'mean(a, axis=None, dtype=None, out=None, keepdims=False)', # 签名
#     {'a': a, 'axis': axis, 'dtype': dtype, 'out': out, 'keepdims': keepdims}, # 变量字典
#     1                                    # 返回值数量
# )
# 
# 而原始逻辑存入 gzipinfo.txt：
# 1. 格式：Gzip 压缩过的 JSON 字符串。
# 2. 内容结构：
#   Key: 函数的路径 ID（如 'numpy.core.numeric.mean'）。
#   Value: 该函数的原始完整源代码字符串。
os.system("python3 Auto_delete_v6.py --dirname {} --path {} --usedfuntionlist {} --unused_gzip_dir {} --builtlist {}".format(dir_name, input_package, used_fun_result_output_final_re, gzip_file, buits_list_file))
```
